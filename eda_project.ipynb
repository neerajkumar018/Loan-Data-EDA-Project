```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# सही फ़ाइल नाम के साथ लोड करें
try:
    df = pd.read_csv("Task 3 and 4_Loan_Data.csv")
    print("✅ Libraries imported and Data loaded successfully!")
    print(f"Dataset Shape: {df.shape[0]} rows and {df.shape[1]} columns\n")
    
    # शुरुआती 5 रोज़ (Rows) देखें
    print("--- First 5 Rows of the Dataset ---")
    display(df.head())
except FileNotFoundError:
    print("❌ Error: फ़ाइल अभी भी नहीं मिली। कृपया नाम दोबारा चेक करें।")
```

```python
# डेटा की स्टेटिस्टिकल समरी (Statistical Summary)
print("--- Statistical Summary ---")
display(df.describe())

# मिसिंग वैल्यूज (Missing Values) चेक करना
print("\n--- Missing Values Count ---")
print(df.isnull().sum())

```

```python
# केवल न्यूमेरिकल कॉलम्स का कोरिलेशन निकालना
numerical_df = df.select_dtypes(include=['float64', 'int64'])
corr = numerical_df.corr()

# हीटमैप विज़ुअलाइज़ेशन बनाना
plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix of Loan Data Features')
plt.show()
plt.show
```
